```python
import pandas as pd

enrolment = pd.read_csv(
    "../data/enrolment.csv",
    low_memory=False
)

print("Enrolment loaded")
```

```python
enrolment.shape
enrolment.head()
```

```python
demographic = pd.read_csv(
    "../data/demographic.csv",
    low_memory=False
)

print("Demographic loaded")
```

```python
demographic.shape
demographic.head()
```

```python
biometric = pd.read_csv(
    "../data/biometric.csv",
    low_memory=False
)

print("Biometric loaded")
```

```python
biometric.shape
biometric.head()
```

```python
enrolment['date'] = pd.to_datetime(enrolment['date'], dayfirst=True)
demographic['date'] = pd.to_datetime(demographic['date'], dayfirst=True)
biometric['date'] = pd.to_datetime(biometric['date'], dayfirst=True)
```

```python
enrolment.info()
```

```python
for df in [enrolment, demographic, biometric]:
    df['year'] = df['date'].dt.year
    df['month'] = df['date'].dt.month
```

```python
enrolment[['date', 'year', 'month']].head()
```

```python
enrolment.isna().sum()
demographic.isna().sum()
biometric.isna().sum()
```

```python
enrolment.to_csv("../data/enrolment_cleaned.csv", index=False)
demographic.to_csv("../data/demographic_cleaned.csv", index=False)
biometric.to_csv("../data/biometric_cleaned.csv", index=False)
```
